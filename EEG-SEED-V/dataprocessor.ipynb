```python
import os
import numpy as np
import matplotlib.pyplot as plt

# Ensure output directory exists
output_dir = "s1s1"
os.makedirs(output_dir, exist_ok=True)

# Define Morlet wavelet function
def morlet_wavelet(t, f0):
    """Generate a Morlet wavelet with center frequency f0."""
    sigma_t = 1 / (2 * np.pi * f0)  # Time standard deviation
    wavelet = np.exp(2j * np.pi * f0 * t) * np.exp(-t**2 / (2 * sigma_t**2))
    return wavelet

# Compute Continuous Wavelet Transform (CWT) manually
def compute_cwt(eeg_signal, sample_rate=200, freqs=np.linspace(1, 40, 50)):
    """
    Compute CWT using Morlet wavelet transform (manual implementation).
    - eeg_signal: (num_channels, num_samples) NumPy array
    - sample_rate: Sampling frequency (default 200 Hz)
    - freqs: Frequencies to analyze (default 1-40 Hz with 50 steps)
    Returns:
        CWT scalogram (num_channels, len(freqs), num_samples)
    """
    num_channels, num_samples = eeg_signal.shape
    time = np.arange(num_samples) / sample_rate  # Time axis
    cwt_output = np.zeros((num_channels, len(freqs), num_samples), dtype=np.float32)

    for ch in range(num_channels):
        for i, f in enumerate(freqs):
            wavelet = morlet_wavelet(time - np.mean(time), f)  # Center wavelet at 0
            conv_result = np.convolve(eeg_signal[ch], np.real(wavelet), mode="same")
            cwt_output[ch, i, :] = np.abs(conv_result)  # Take magnitude

    return cwt_output

# Process each trial and band
for trial_idx, trial_data in enumerate(filtered_trials):
    for band_idx, (band_name, band_data) in enumerate(trial_data.items()):
        # Compute CWT for this band
        cwt_image = compute_cwt(np.array(band_data))  # Convert to NumPy array

        # Save each channel as an image
        for ch in range(cwt_image.shape[0]):
            fig, ax = plt.subplots(figsize=(8, 6))
            ax.imshow(cwt_image[ch], aspect="auto", cmap="jet", origin="lower")
            ax.set_xticks([])  # Remove x-axis ticks
            ax.set_yticks([])  # Remove y-axis ticks
            ax.set_frame_on(False)  # Remove border

            # Save image
            img_path = os.path.join(output_dir, f"trial_{trial_idx+1}_band_{band_idx+1}_ch_{ch+1}.png")
            plt.savefig(img_path, dpi=300, bbox_inches='tight', pad_inches=0)
            plt.close()

print(f"All CWT images saved in '{output_dir}'")

```
